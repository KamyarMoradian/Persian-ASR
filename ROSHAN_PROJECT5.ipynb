# Evaluation

## Utils

```python
def normalize_text(text):
    # Step 0: Remove any space from beginning and end of sentence
    text = text.strip()

    # Step 1: Use Parsivar to normalize data
    normalizer = Normalizer()
    text = normalizer.normalize(text)

    # Step 2: Remove letters repeated more than twice
    text = re.sub(r'(.)\1{2,}', r'\1', text)

    # Step 3: Remove any Arabic diacritics from the text
    arabic_diacritics = re.compile("""
                             ّ    | # Tashdid
                             َ    | # Fatha
                             ً    | # Tanwin Fath
                             ُ    | # Damma
                             ٌ    | # Tanwin Damm
                             ِ    | # Kasra
                             ٍ    | # Tanwin Kasr
                             ْ    | # Sukun
                             ـ     # Tatwil/Kashida
                         """, re.VERBOSE)
    text = re.sub(arabic_diacritics, '', text)

    # Step 4: Remove the hashtag sign (“#”) from the text while keeping the information included in the hashtags
    text = re.sub(r'#', ' ', text)
    text = text.replace('_', ' ')

    # Step 5: Last normalization
    text = normalizer.normalize(text)

    return text
```

```python
# Function to evaluate a single transcription
def evaluate_sentence(reference, hypothesis):
    word_error_rate = wer(reference, hypothesis)
    char_error_rate = cer(reference, hypothesis)
    return word_error_rate, char_error_rate


# Evaluate the model on the test dataset
def evaluate(eval_ds, transcriber):
    total_wer = 0
    total_cer = 0
    num_samples = len(eval_ds)

    for i in tqdm(range(num_samples)):
        sample = eval_ds[i]
        audio, reference_transcript = sample['audio'], sample['sentence']
        hypothesis_transcript = transcriber(audio)
        sample_wer, sample_cer = evaluate_sentence(
            normalize_text(reference_transcript),
            normalize_text(hypothesis_transcript)
        )
        total_wer += sample_wer
        total_cer += sample_cer

    avg_wer = total_wer / num_samples
    avg_cer = total_cer / num_samples

    return {
        'WER': avg_wer,
        'CER': avg_cer
    }
```

```python
import soundfile as sf

def save_audio(out_dir, audio_name, audio, sampling_rate):
    out_path = os.path.join(out_dir, audio_name)
    sf.write(out_path, audio, sampling_rate)
    return out_path
```

```python
import unicodedata
import string

SKIP = set(
    list(string.ascii_letters)
    + [
        "=",  # occurs only 2x in utterance (transl.): "twenty = xx"
        "ā",  # occurs only 4x together with "š"
        "š",
        # Arabic letters
        "ة",  # TEH MARBUTA
    ]
)

DISCARD = [
    # "(laughter)" in Farsi
    "(خنده)",
    # ASCII
    "!",
    '"',
    "#",
    "&",
    "'",
    "(",
    ")",
    ",",
    "-",
    ".",
    ":",
    ";",
    # Unicode punctuation?
    "–",
    "“",
    "”",
    "…",
    "؟",
    "،",
    "؛",
    "ـ",
    # Unicode whitespace?
    "ً",
    "ٌ",
    "َ",
    "ُ",
    "ِ",
    "ّ",
    "ْ",
    "ٔ",
    # Other
    "«",
    "»",
    "?",
    '⁇',
    ",",
    ">",
    "<"
]

REPLACEMENTS = {
    "أ": "ا",
    "ۀ": "ە",
    "ك": "ک",
    "ي": "ی",
    "ى": "ی",
    "ﯽ": "ی",
    "ﻮ": "و",
    "ے": "ی",
    "ﺒ": "ب",
    "ﻢ": "ﻡ",
    "٬": " ",
    "ە": "ه",
}


def maybe_normalize(text: str) -> str | None:

    # Skip selected with banned characters
    if set(text) & SKIP:
        return None  # skip this

    # Remove hashtags - they are not being read in Farsi CV
    text = " ".join(w for w in text.split() if not w.startswith("#"))

    # Replace selected characters with others
    for lhs, rhs in REPLACEMENTS.items():
        text = text.replace(lhs, rhs)

    # Replace selected characters with empty strings
    for tok in DISCARD:
        text = text.replace(tok, "")

    # Unify the symbols that have the same meaning but different Unicode representation.
    text = unicodedata.normalize("NFKC", text)

    # Remove hamza's that were not merged with any letter by NFKC.
    text = text.replace("ء", "")

    # Remove double whitespace etc.
    return " ".join(t for t in text.split() if t)
```

```python
def show_samples_predicted(transcriber):
    for i in SAMPLE_INDICES_TO_SHOW:
        print(f"SAMPLE {i}:")
        print("MAIN SENTENCE:")
        print(f'\t - {maybe_normalize(train_ds[i]["sentence"])}')
        print("PREDICTED SENTENCE:")
        print(f'\t - {transcriber(train_ds[i]["audio"])}')
        print()
```

```python
def seamless_transcriber(audio):
    audio_inputs = seamless_processor(
        audios=audio['array'],
        return_tensors="pt",
        sampling_rate=16000
    ).to(device)

    output_tokens = seamless_model.generate(
        **audio_inputs,
        tgt_lang='pes',
        generate_speech=False
    )

    translated_text_from_audio = seamless_processor.decode(
        output_tokens[0].tolist()[0],
        skip_special_tokens=True,
    )

    return translated_text_from_audio
```

```python
# --- Building Manifest Files --- #
import json

# Function to build a manifest
def build_manifest(ds, manifest_path, wav_path):
    with open(manifest_path, 'w', encoding='utf-8') as fout:
        for row in ds:
            sentence = row['sentence'].strip()
            audio_name = row['audio']['path'].split('.')[0]
            audio_path = os.path.join(wav_path, audio_name + '.wav')

            duration = librosa.get_duration(
                y=row['audio']['array'],
                sr=row['audio']['sampling_rate']
            )

            metadata = {
                "audio_filepath": audio_path,
                "duration": duration,
                "text": maybe_normalize(sentence)
            }

            if metadata['text'] == None:
                continue

            json.dump(metadata, fout, ensure_ascii=False)
            fout.write('\n')

# Building Manifests
print("******")
train_manifest = os.path.join(data_dir, 'train_manifest.json')
if not os.path.isfile(train_manifest):
    build_manifest(train_eval_ds['train'], train_manifest, 'train_wav')
    print("Training manifest created.")

test_manifest = os.path.join(data_dir, 'test_manifest.json')
if not os.path.isfile(test_manifest):
    build_manifest(train_eval_ds['test'], test_manifest, 'test_wav')
    print("Testing manifest created.")
print("***Done***")
```

## Load and setup training

```python
import nemo.collections.asr as nemo_asr
nemo_asr_model = nemo_asr.models.EncDecHybridRNNTCTCBPEModel.from_pretrained(
    model_name="nvidia/stt_fa_fastconformer_hybrid_large"
).to(device)
```

```python
nemo_asr_model.setup_training_data(train_data_config={
    "manifest_filepath": train_manifest,
    "batch_size": 8,
    "num_workers": 2,
    "sample_rate": 16000,
    "pin_memory": True,
    "trim_silence": True,
    "normalize_transcripts": False
})

nemo_asr_model.setup_validation_data(val_data_config={
    "manifest_filepath": test_manifest,
    "batch_size": 8,
    "num_workers": 2,
    "sample_rate": 16000,
    "pin_memory": True,
    "trim_silence": True,
    "normalize_transcripts": False
})

nemo_asr_model.setup_test_data(test_data_config={
    "manifest_filepath": test_manifest,
    "batch_size": 8,
    "num_workers": 2,
    "sample_rate": 16000,
    "pin_memory": True,
    "trim_silence": True,
    "normalize_transcripts": False
})
```

```python
nemo_asr_model.cfg.optim.sched.warmup_steps = None
nemo_asr_model.cfg.optim.sched.warmup_ratio = 0.1
```

```python
nemo_asr_model.spec_augmentation = nemo_asr_model.from_config_dict(nemo_asr_model.cfg.spec_augment)
```

```python
accelerator = 'gpu' if torch.cuda.is_available() else 'cpu'

trainer = pl.Trainer(
    accelerator=accelerator,
    devices=-1,
    max_epochs=10,
    strategy='auto',
    val_check_interval=1.0,
    precision=16,
    accumulate_grad_batches=4,
    enable_checkpointing=False,
    logger=False,
    log_every_n_steps=20,
    check_val_every_n_epoch=1
)
```
